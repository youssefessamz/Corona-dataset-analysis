```python
df2["AGE_GROUP"] = pd.cut(
    df2["AGE"],
    bins=[-1, 18, 30, 45, 60, 74, np.inf],
    labels=["0-18", "19-30", "31-45", "46-60", "61-74", "75+"]
)
```

```python
df2.groupby("AGE_GROUP")["PATIENT_TYPE"].apply(lambda s: (s == 2).mean() * 100)
```

```python
ax = df2.groupby("AGE_GROUP")["PATIENT_TYPE"].apply(
    lambda s: (s == 2).mean() * 100
).plot(kind="bar")

ax.set_ylabel("Hospitalization rate (%)")
ax.yaxis.set_major_formatter(lambda value, position: f"{value:.0f}%")
```

```python
diabetes_and_hypertension_rate = (df2.loc[(df2["DIABETES"] == 1) &(df2["HIPERTENSION"] == 1), "PATIENT_TYPE"].eq(2).mean() * 100)
print(diabetes_and_hypertension_rate)
diabetes_and_hypertension_rate_negative = (df2.loc[(df2["DIABETES"] == 2) &(df2["HIPERTENSION"] == 2), "PATIENT_TYPE"].eq(2).mean() * 100)
print(diabetes_and_hypertension_rate_negative)
```

```python
condition_cols = ["DIABETES", "HIPERTENSION", "OBESITY", "ASTHMA", "COPD", "CARDIOVASCULAR"]
df2["CONDITION_COUNT"] = df2[condition_cols].eq(1).sum(axis=1)
```

```python
condition_summary = df2["CONDITION_COUNT"].value_counts().sort_index().to_frame("patients")
condition_summary["percent"] = (
    condition_summary["patients"] / len(df2) * 100
)

print(condition_summary.round(2))
```

```python
df2.loc[df2["PATIENT_TYPE"] == 2, "CONDITION_COUNT"].eq(0).mean()*100
```

```python
hospitalized_zero_chronicle_condition_rate = (df2.loc[df2["CONDITION_COUNT"] == 0, "PATIENT_TYPE"] == 2).mean() * 100
hospitalized_chronicle_condition_rate= (df2.loc[df2["CONDITION_COUNT"] != 0, "PATIENT_TYPE"] == 2).mean() * 100
print(hospitalized_zero_chronicle_condition_rate)
print(hospitalized_chronicle_condition_rate)
```

```python
columns = ["PNEUMONIA","PREGNANT", "DIABETES", "COPD", "ASTHMA", "INMSUPR", "HIPERTENSION"]
df2.groupby("PATIENT_TYPE")[columns].agg(lambda s: s.eq(1).sum())
```

```python
columns = ["PNEUMONIA","PREGNANT", "DIABETES", "COPD", "ASTHMA", "INMSUPR", "HIPERTENSION"]
for col in columns:
    valid = df2[df2[col].isin([1, 2])]
    rates = pd.crosstab(
        valid[col], valid["PATIENT_TYPE"], normalize="index"
    ).mul(100).round(2)
    print(f"\n{col}:")
    print(rates)
```

```python
deaths_1_to_3 = df2[df2["DATE_DIED"].ne("9999-99-99")& df2["CLASIFFICATION_FINAL"].between(1, 3)]
counts = deaths_1_to_3["CLASIFFICATION_FINAL"].value_counts().sort_index()
print(counts)
print("Total:", counts.sum())
```
